```python
import importlib
import importlib.metadata as metadata
import subprocess
import sys

if sys.version_info < (3, 10):
    raise RuntimeError("Selecione um kernel Python 3.10 ou superior.")

# Mesmos intervalos de requirements.txt. O notebook é autossuficiente no Colab.
REQUISITOS = {
    "cryptography": ">=50.0.1,<51",
    "ipywidgets": ">=8.1.9,<9",
    "zxcvbn": ">=4.5.0,<5",
    "ipython": ">=8.18,<10",
    "packaging": ">=24,<27",
}

# packaging é usado diretamente para interpretar versões e intervalos (PEP 440).
try:
    from packaging.specifiers import SpecifierSet
except ImportError:
    subprocess.check_call([sys.executable, "-m", "pip", "install", "--disable-pip-version-check",
                           "packaging" + REQUISITOS["packaging"]])
    importlib.invalidate_caches()
    from packaging.specifiers import SpecifierSet

def dependencias_pendentes(requisitos, consultar=metadata.version):
    pendentes = []
    for nome, intervalo in requisitos.items():
        try:
            instalada = consultar(nome)
        except metadata.PackageNotFoundError:
            instalada = None
        if instalada is None or not SpecifierSet(intervalo).contains(instalada, prereleases=False):
            pendentes.append(nome + intervalo)
    return pendentes

pendentes = dependencias_pendentes(REQUISITOS)
modulos_antes = set(sys.modules)
if pendentes:
    print("Instalando ou ajustando dependências:", ", ".join(pendentes))
    subprocess.check_call([sys.executable, "-m", "pip", "install", "--disable-pip-version-check", *pendentes])
    importlib.invalidate_caches()
if dependencias_pendentes(REQUISITOS):
    raise RuntimeError("Dependências fora do intervalo. Confira a saída da instalação e o kernel selecionado.")

# Bibliotecas já carregadas não podem ser trocadas com segurança dentro do mesmo processo.
modulos = {"ipython": "IPython", "cryptography": "cryptography", "ipywidgets": "ipywidgets",
           "zxcvbn": "zxcvbn", "packaging": "packaging"}
alteradas_carregadas = [nome for nome in REQUISITOS
    if nome + REQUISITOS[nome] in pendentes and modulos[nome] in modulos_antes]
if alteradas_carregadas:
    raise RuntimeError("Instalação concluída. Reinicie o kernel/sessão e execute tudo para carregar: "
                       + ", ".join(alteradas_carregadas))

import base64
import hashlib
import hmac
import json
import secrets
import tempfile
from pathlib import Path

import cryptography
import ipywidgets as widgets
from IPython.display import display
from cryptography.exceptions import InvalidSignature, InvalidTag
from cryptography.hazmat.primitives import hashes, serialization
from cryptography.hazmat.primitives.asymmetric import padding, rsa
from cryptography.hazmat.primitives.ciphers.aead import AESGCM

def configurar_widgets_colab():
    # Somente no Colab; não instalar google-colab em Jupyter/VS Code.
    try:
        import google.colab
    except ImportError:
        return "Jupyter/VS Code: widgets padrão disponíveis no kernel."
    from google.colab import output
    output.enable_custom_widget_manager()
    return "Google Colab: gerenciador de widgets habilitado."

PASTA_LAB = Path(tempfile.mkdtemp(prefix="criptolab_"))
print("Python:", sys.version.split()[0])
for nome in REQUISITOS:
    print(f"{nome}: {metadata.version(nome)}")
print(configurar_widgets_colab())
print("Setup concluído. Arquivos didáticos usarão uma pasta temporária exclusiva.")
```

```python
# Estas funções tornam os testes negativos explícitos, sem esconder erros inesperados.
TESTES = []

def confirmar(condicao, descricao):
    if not condicao:
        raise AssertionError(descricao)
    TESTES.append(descricao)
    print("✅", descricao)

def esperar_erro(tipo, acao, descricao):
    try:
        acao()
    except tipo:
        TESTES.append(descricao)
        print(f"✅ {descricao} ({tipo.__name__})")
    else:
        raise AssertionError("Falha esperada não ocorreu: " + descricao)

def alterar_um_bit(dados):
    # Trabalhamos em uma cópia, preservando os bytes originais.
    if not dados:
        raise ValueError("Forneça ao menos um byte.")
    copia = bytearray(dados)
    copia[0] ^= 1
    return bytes(copia)
```

### Motor dos painéis

Execute a próxima célula uma vez. Ela define a interface; os painéis aparecem nos blocos abaixo.
O código pode ser recolhido para focar na atividade. Mensagens usam até 16 KiB por painel;
RSA direto aceita 190 bytes e o estimador de senhas, até 72 caracteres, sem truncamento.


```python
# Infraestrutura dos painéis demonstrativos e conexão com a implementação da atividade final.
import html
import re
import ipywidgets as widgets
from IPython.display import display
from zxcvbn import zxcvbn


def avaliar_forca(senha, minimo=15):
    """Retorna apenas diagnóstico; nunca devolve a senha nem fragmentos dela."""
    if not senha:
        raise ValueError("Digite uma senha fictícia antes de avaliar.")
    if len(senha) > 72:
        raise ValueError("O estimador deste laboratório aceita até 72 caracteres; nada foi truncado.")
    resultado = zxcvbn(senha, user_inputs=[
        "oxetech", "securebox", "criptolab", "senha", "ciberseguranca", "laboratorio"])
    score = resultado["score"]
    padroes = {trecho["pattern"] for trecho in resultado["sequence"]}
    motivos = []
    if len(senha) < minimo:
        motivos.append(f"Comprimento abaixo do mínimo didático escolhido: {minimo} caracteres.")
    nomes = {"dictionary": "Palavras ou combinações reconhecidas pelo estimador.",
             "repeat": "Repetições previsíveis.", "sequence": "Sequências previsíveis.",
             "spatial": "Caminhos de teclado.", "date": "Padrões de data.",
             "regex": "Padrão numérico reconhecido, como um ano."}
    motivos.extend(nomes[p] for p in sorted(padroes) if p in nomes)
    forte = score >= 3 and len(senha) >= minimo
    classificacao = "Forte (estimativa)" if forte else ("Fraca" if score <= 1 else "Precisa melhorar")
    sugestoes = ["Prefira uma frase-senha longa com palavras escolhidas ao acaso ou um gerador de senhas.",
                "Use uma senha exclusiva por serviço; não use os exemplos da aula em contas reais."]
    if score >= 3 and not forte:
        sugestoes.insert(0, "O estimador gostou da imprevisibilidade; aumente o comprimento para cumprir o critério da turma.")
    # Não devolvemos o resultado bruto do zxcvbn: ele inclui a senha e trechos reconhecidos.
    return {"classificacao": classificacao, "score": score, "comprimento": len(senha),
            "forte": forte, "motivos": motivos, "sugestoes": sugestoes}


class PainelCriptoLab:
    """Um painel por atividade; estado isolado e callbacks sem arquivos ou chamadas de rede."""

    def __init__(self, modo):
        self.modo = modo
        self.estado = {}
        self.botoes = {}
        self.campos = {}
        self.erro_inesperado = None
        self.status = widgets.HTML(value="<p>Escolha uma ação para começar.</p>")
        self.resultado = widgets.Textarea(value="", disabled=True,
                                          layout=widgets.Layout(width="100%", height="170px"))
        self.conteudo = []
        self.configurar()
        acoes = widgets.Box(list(self.botoes.values()),
                            layout=widgets.Layout(display="flex", flex_flow="row wrap", gap="6px"))
        self.widget = widgets.VBox(self.conteudo + [acoes, self.status, self.resultado],
                                  layout=widgets.Layout(width="100%", border="1px solid #9ca3af", padding="14px"))

    def texto(self, nome, legenda, valor="", grande=True):
        self.conteudo.append(widgets.HTML(value=f"<b>{html.escape(legenda)}</b>"))
        campo = (widgets.Textarea(value=valor, layout=widgets.Layout(width="100%", height="90px"))
                 if grande else widgets.Text(value=valor, layout=widgets.Layout(width="100%")))
        self.campos[nome] = campo
        self.conteudo.append(campo)
        return campo

    def botao(self, nome, legenda, acao, estilo=""):
        botao = widgets.Button(description=legenda, button_style=estilo,
                               layout=widgets.Layout(width="auto", min_width="120px"))

        def clicar(_):
            self.erro_inesperado = None
            self.resultado.value = ""  # Nunca deixe um sucesso antigo visível após uma falha.
            botao.disabled = True
            try:
                acao()
            except NotImplementedError:
                self.mostrar("ATIVIDADE PENDENTE — complete a implementação da missão e execute os testes.")
            except (InvalidTag, InvalidSignature):
                self.mostrar("RECUSADO — a autenticação não passou. Nenhuma mensagem foi liberada.", erro=True)
            except (ValueError, KeyError) as exc:
                self.mostrar(str(exc), erro=True)
            except Exception:
                # Não ecoar entradas de usuário em exceções; sinalizar falha de implementação.
                self.erro_inesperado = True
                self.mostrar("Erro inesperado. Reinicie a sessão e informe o docente.", erro=True)
                raise
            finally:
                botao.disabled = False
        botao.on_click(clicar)
        self.botoes[nome] = botao

    def mostrar(self, mensagem, detalhe="", erro=False):
        cor = "#991b1b" if erro else "#166534"
        self.status.value = f'<p style="color:{cor}"><b>{html.escape(mensagem)}</b></p>'
        self.resultado.value = detalhe

    def exigir(self, chave, mensagem):
        if chave not in self.estado:
            raise ValueError(mensagem)
        return self.estado[chave]

    def bytes_mensagem(self):
        dados = self.campos["mensagem"].value.encode("utf-8")
        if len(dados) > 16_384:
            raise ValueError("Use até 16 KiB de texto neste painel didático.")
        return dados

    def limpar(self):
        self.estado.clear()
        for campo in self.campos.values():
            if isinstance(campo, (widgets.Text, widgets.Textarea, widgets.Password)):
                campo.value = ""
            elif isinstance(campo, widgets.Dropdown):
                campo.index = 0
        self.mostrar("Painel reiniciado. Referências e chaves anteriores foram descartadas da atividade.")

    def configurar(self):
        if self.modo == "fundamentos":
            perguntas = [("leitura", "Uma pessoa indevida leu o boletim."),
                         ("mudanca", "O horário foi alterado durante a transmissão."),
                         ("origem", "Precisamos conferir se veio da equipe de campo.")]
            for nome, pergunta in perguntas:
                self.conteudo.append(widgets.HTML(value=html.escape(pergunta)))
                campo = widgets.Dropdown(options=["Escolha…", "Confidencialidade", "Integridade", "Autenticidade"],
                                         layout=widgets.Layout(width="100%"))
                self.campos[nome] = campo
                self.conteudo.append(campo)
            self.botao("responder", "Registrar respostas", self.responder, "primary")
        elif self.modo == "senha":
            self.conteudo.append(widgets.HTML(value="<b>Digite apenas senhas fictícias</b>"))
            self.campos["senha"] = widgets.Password(placeholder="Senha de teste", continuous_update=False,
                layout=widgets.Layout(width="100%"))
            self.campos["minimo"] = widgets.IntSlider(value=15, min=8, max=24, description="Mínimo:",
                continuous_update=False, style={"description_width": "initial"})
            self.conteudo.extend([self.campos["senha"], self.campos["minimo"]])
            self.botao("avaliar", "Avaliar força", self.avaliar, "primary")
            self.botao("cadastrar", "Cadastrar teste", self.cadastrar)
            self.botao("verificar", "Conferir senha", self.conferir)
        elif self.modo == "hash":
            self.texto("mensagem", "Mensagem original", "Sensor S-17: manutenção às 14h.")
            self.texto("comparacao", "Mensagem recebida / comparação", "Sensor S-17: manutenção às 18h.")
            self.botao("hash", "Gerar SHA-256", self.gerar_hash, "primary")
            self.botao("comparar", "Comparar mensagens", self.comparar_hash)
            self.botao("guardar", "Guardar referência", self.guardar_hash)
            self.botao("integridade", "Verificar recebida", self.conferir_hash)
        else:
            self.texto("mensagem", "Mensagem para experimentar", "Olá, oficina! Teste do OxeTech Lab. 🔐")
            if self.modo in ("aes", "missao"):
                self.texto("aad", "Contexto esperado (AAD público)", "SecureBox|turma|v1", grande=False)
            if self.modo == "aes":
                self.botao("chave", "Nova chave", self.nova_aes)
                self.botao("cifrar", "Criptografar", self.cifrar_aes, "primary")
                self.botao("decifrar", "Descriptografar", self.decifrar_aes)
                self.botao("adulterar", "Adulterar cópia", self.adulterar_aes)
                self.botao("errada", "Testar chave errada", self.errada_aes)
            elif self.modo == "rsa":
                self.botao("chaves", "Gerar par RSA", self.nova_rsa)
                self.botao("pem", "Ver PEM público", self.publico_pem)
                self.botao("cifrar", "Criptografar", self.cifrar_rsa, "primary")
                self.botao("decifrar", "Descriptografar", self.decifrar_rsa)
            elif self.modo == "assinatura":
                self.botao("chaves", "Gerar par de assinatura", self.nova_rsa)
                self.botao("assinar", "Assinar mensagem", self.assinar, "primary")
                self.botao("validar", "Validar texto atual", self.validar)
                self.botao("adulterar", "Testar adulteração", self.adulterar_assinatura)
            elif self.modo == "missao":
                self.botao("enviar", "Proteger e enviar", self.enviar, "primary")
                self.botao("receber", "Receber na oficina", self.receber)
                self.botao("adulterar", "Enviar cópia adulterada", self.adulterar_missao)
        self.botao("limpar", "Reiniciar painel", self.limpar)

    def responder(self):
        if any(self.campos[nome].value == "Escolha…" for nome in ("leitura", "mudanca", "origem")):
            raise ValueError("Responda às três situações antes de registrar.")
        self.mostrar("Respostas registradas para discussão. Justifique suas escolhas ao docente.")

    def avaliar(self):
        try:
            diagnostico = avaliar_forca(self.campos["senha"].value, self.campos["minimo"].value)
        finally:
            self.campos["senha"].value = ""
        detalhe = (f"Comprimento: {diagnostico['comprimento']} | Pontuação estimada: {diagnostico['score']}/4\n\n"
                   + "\n".join(diagnostico["motivos"] or ["Nenhum padrão destacado pelo estimador."])
                   + "\n\n" + "\n".join(diagnostico["sugestoes"])
                   + "\n\nEstimativa, não garantia. Não consulta vazamentos; dicionários têm cobertura limitada em português.")
        self.mostrar(diagnostico["classificacao"], detalhe, erro=not diagnostico["forte"])

    def cadastrar(self):
        # Aceitamos senhas fracas somente nesta simulação, para comparar força e armazenamento.
        try:
            senha = self.campos["senha"].value
            if not senha or len(senha) > 72:
                raise ValueError("Digite uma senha fictícia entre 1 e 72 caracteres.")
            salt = secrets.token_bytes(16)
            verificador = hashlib.pbkdf2_hmac("sha256", senha.encode("utf-8"), salt, 600_000)
            self.estado["cadastro"] = {"salt": salt, "verificador": verificador, "iteracoes": 600_000}
        finally:
            self.campos["senha"].value = ""
        self.mostrar("Cadastro de teste criado. Digite novamente para conferir.",
                     "Guardados em memória: salt, custo e verificador PBKDF2.\n"
                     "Este botão não aprova a força da senha; avalie-a separadamente.")

    def conferir(self):
        try:
            registro = self.exigir("cadastro", "Crie um cadastro de teste primeiro.")
            senha = self.campos["senha"].value
            if not senha or len(senha) > 72:
                raise ValueError("Digite uma senha fictícia entre 1 e 72 caracteres.")
            verificador = hashlib.pbkdf2_hmac("sha256", senha.encode("utf-8"), registro["salt"], registro["iteracoes"])
            aceita = hmac.compare_digest(verificador, registro["verificador"])
        finally:
            self.campos["senha"].value = ""
        self.mostrar("Senha confere com o cadastro." if aceita else "Senha não confere com o cadastro.", erro=not aceita)

    def gerar_hash(self):
        dados = self.bytes_mensagem()
        self.mostrar("SHA-256 calculado.", hashlib.sha256(dados).hexdigest())

    def comparar_hash(self):
        a = self.bytes_mensagem()
        b = self.campos["comparacao"].value.encode("utf-8")
        if len(b) > 16_384:
            raise ValueError("Use até 16 KiB na comparação.")
        ha, hb = hashlib.sha256(a).digest(), hashlib.sha256(b).digest()
        bits = sum((x ^ y).bit_count() for x, y in zip(ha, hb))
        self.mostrar("Mensagens iguais." if a == b else "Mensagens diferentes.",
            f"Original: {ha.hex()}\nRecebida: {hb.hex()}\nBits diferentes nos hashes: {bits}/256 ({bits/256:.1%})")

    def guardar_hash(self):
        self.estado["referencia"] = hashlib.sha256(self.bytes_mensagem()).hexdigest()
        self.mostrar("Referência do original guardada nesta sessão.", self.estado["referencia"])

    def conferir_hash(self):
        referencia = self.exigir("referencia", "Guarde a referência da mensagem original primeiro.")
        atual = hashlib.sha256(self.campos["comparacao"].value.encode("utf-8")).hexdigest()
        confere = hmac.compare_digest(referencia, atual)
        self.mostrar("Integridade confirmada." if confere else "Alteração detectada.", erro=not confere)

    def nova_aes(self):
        self.estado.clear()
        self.estado["chave"] = AESGCM.generate_key(bit_length=256)
        self.mostrar("Nova chave AES criada. O pacote anterior foi descartado.")

    def cifrar_aes(self):
        dados = self.bytes_mensagem()
        if "chave" not in self.estado:
            self.estado["chave"] = AESGCM.generate_key(bit_length=256)
        nonce = secrets.token_bytes(12)
        aad = self.campos["aad"].value.encode("utf-8")
        cifra = AESGCM(self.estado["chave"]).encrypt(nonce, dados, aad)
        self.estado["pacote"] = {"nonce": nonce, "cifra": cifra, "aad": aad}
        self.mostrar("Mensagem criptografada; cada clique usa um novo nonce.",
            f"Nonce: {nonce.hex()}\nCifra + tag (Base64):\n{base64.b64encode(cifra).decode('ascii')}")

    def abrir_aes(self, pacote, chave):
        aad = self.campos["aad"].value.encode("utf-8")
        dados = AESGCM(chave).decrypt(pacote["nonce"], pacote["cifra"], aad)
        self.mostrar("Autenticação válida. Mensagem recuperada:", dados.decode("utf-8"))

    def decifrar_aes(self):
        self.abrir_aes(self.exigir("pacote", "Criptografe uma mensagem primeiro."), self.estado["chave"])

    def adulterar_aes(self):
        pacote = dict(self.exigir("pacote", "Criptografe uma mensagem primeiro."))
        pacote["cifra"] = alterar_um_bit(pacote["cifra"])
        self.abrir_aes(pacote, self.estado["chave"])

    def errada_aes(self):
        self.abrir_aes(self.exigir("pacote", "Criptografe uma mensagem primeiro."), AESGCM.generate_key(bit_length=256))

    def nova_rsa(self):
        self.estado.clear()
        privada = rsa.generate_private_key(public_exponent=65537, key_size=2048)
        self.estado.update(privada=privada, publica=privada.public_key())
        self.mostrar("Novo par RSA gerado. Cifras ou assinaturas anteriores foram descartadas.")

    def garantir_rsa(self):
        if "privada" not in self.estado:
            self.nova_rsa()

    def publico_pem(self):
        self.garantir_rsa()
        pem = self.estado["publica"].public_bytes(serialization.Encoding.PEM,
                                                serialization.PublicFormat.SubjectPublicKeyInfo)
        self.mostrar("Chave pública em PEM. A privada permanece na sessão.", pem.decode("ascii"))

    def cifrar_rsa(self):
        dados = self.bytes_mensagem()
        if len(dados) > 190:
            raise ValueError(f"Mensagem com {len(dados)} bytes: o limite RSA-OAEP deste painel é 190. Use a missão híbrida para textos maiores.")
        self.garantir_rsa()
        esquema = padding.OAEP(mgf=padding.MGF1(hashes.SHA256()), algorithm=hashes.SHA256(), label=None)
        self.estado["cifra"] = self.estado["publica"].encrypt(dados, esquema)
        self.mostrar("RSA-OAEP aplicado à mensagem.", base64.b64encode(self.estado["cifra"]).decode("ascii"))

    def decifrar_rsa(self):
        cifra = self.exigir("cifra", "Criptografe uma mensagem primeiro.")
        esquema = padding.OAEP(mgf=padding.MGF1(hashes.SHA256()), algorithm=hashes.SHA256(), label=None)
        dados = self.estado["privada"].decrypt(cifra, esquema)
        self.mostrar("Mensagem RSA recuperada:", dados.decode("utf-8"))

    def assinar(self):
        dados = self.bytes_mensagem()
        self.garantir_rsa()
        esquema = padding.PSS(mgf=padding.MGF1(hashes.SHA256()), salt_length=padding.PSS.MAX_LENGTH)
        self.estado["assinatura"] = self.estado["privada"].sign(dados, esquema, hashes.SHA256())
        self.mostrar("Mensagem assinada. Agora altere o texto e valide novamente.",
                     base64.b64encode(self.estado["assinatura"]).decode("ascii"))

    def conferir_assinatura(self, dados):
        assinatura = self.exigir("assinatura", "Assine uma mensagem primeiro.")
        esquema = padding.PSS(mgf=padding.MGF1(hashes.SHA256()), salt_length=padding.PSS.MAX_LENGTH)
        self.estado["publica"].verify(assinatura, dados, esquema, hashes.SHA256())
        self.mostrar("Assinatura válida para o texto atual e a chave desta sessão.")

    def validar(self):
        self.conferir_assinatura(self.bytes_mensagem())

    def adulterar_assinatura(self):
        self.conferir_assinatura(self.bytes_mensagem() + b"!")

    def enviar(self):
        dados = self.bytes_mensagem()
        if "destino" not in self.estado:
            self.estado["destino"] = gerar_par_rsa()
            self.estado["remetente"] = gerar_par_rsa()
        aad = self.campos["aad"].value.encode("utf-8")
        envelope, assinatura = construir_envelope(dados.decode("utf-8"), aad,
            self.estado["destino"][1], self.estado["remetente"][0])
        self.estado.update(envelope=envelope, assinatura=assinatura)
        self.mostrar("Envelope enviado à oficina simulada.", bytes_envelope(envelope).decode("utf-8"))

    def abrir_envelope(self, envelope):
        dados = receber_envelope(envelope, self.estado["assinatura"], self.estado["remetente"][1],
            self.estado["destino"][0], self.campos["aad"].value.encode("utf-8"))
        self.mostrar("Oficina: assinatura e autenticação válidas. Mensagem recebida:", dados)

    def receber(self):
        self.abrir_envelope(self.exigir("envelope", "Proteja e envie uma mensagem primeiro."))

    def adulterar_missao(self):
        envelope = dict(self.exigir("envelope", "Proteja e envie uma mensagem primeiro."))
        envelope["cifra"] = alterar_um_bit(envelope["cifra"])
        self.abrir_envelope(envelope)
```

## Bloco 1 — Hashing: SHA-256, salt e comparação de hashes

### 📖 CONCEITO

SHA-256 produz um resumo de 256 bits, normalmente mostrado em 64 caracteres hexadecimais.
O mesmo conteúdo produz o mesmo resumo. Mudanças pequenas tendem a mudar muitos bits do
resultado: o **efeito avalanche**. Não esperamos exatamente 50% em cada comparação.

Um hash não permite “decifrar” o original. Como o espaço de entradas é maior que o de saídas,
colisões existem; o objetivo de segurança é tornar inviável encontrar colisões úteis.
Para conferir um arquivo, precisamos também confiar no hash de referência.

### 🧪 LABORATÓRIO — uma mudança de um bit


```python
def hash_bytes(dados):
    return hashlib.sha256(dados).hexdigest()

original_hash = boletim_bytes
modificado_hash = alterar_um_bit(original_hash)
resumo_a = hashlib.sha256(original_hash).digest()
resumo_b = hashlib.sha256(modificado_hash).digest()
bits_entrada = sum((a ^ b).bit_count() for a, b in zip(original_hash, modificado_hash))
bits_saida = sum((a ^ b).bit_count() for a, b in zip(resumo_a, resumo_b))
print("SHA-256 original: ", resumo_a.hex())
print("SHA-256 alterado: ", resumo_b.hex())
print(f"Entrada: {bits_entrada} bit diferente | Saída: {bits_saida}/256 ({bits_saida/256:.1%})")
confirmar(bits_entrada == 1, "A entrada mudou exatamente um bit")
confirmar(resumo_a != resumo_b, "A mudança deste exemplo alterou o hash")
confirmar(hash_bytes(original_hash) == hash_bytes(original_hash), "SHA-256 é determinístico")
```

### 🧪 LABORATÓRIO — integridade de arquivo

Vamos tratar o hash registrado antes da alteração como uma referência confiável **apenas neste
experimento**. Em um download real, guardar arquivo e hash em um canal comprometido não basta.
Ler em blocos permite calcular o hash sem carregar um arquivo grande inteiro na memória.


```python
def hash_arquivo(caminho):
    resumo = hashlib.sha256()
    with Path(caminho).open("rb") as arquivo:
        for bloco in iter(lambda: arquivo.read(65536), b""):
            resumo.update(bloco)
    return resumo.hexdigest()

arquivo_boletim = PASTA_LAB / "boletim.txt"
arquivo_boletim.write_bytes(boletim_bytes)
referencia_confiavel = hash_arquivo(arquivo_boletim)
confirmar(hmac.compare_digest(hash_arquivo(arquivo_boletim), referencia_confiavel),
          "Arquivo original coincide com a referência")
arquivo_boletim.write_bytes(boletim_bytes + b" Revisao nao autorizada.")
confirmar(not hmac.compare_digest(hash_arquivo(arquivo_boletim), referencia_confiavel),
          "Alteração do arquivo foi detectada")
arquivo_boletim.write_bytes(boletim_bytes)  # Restaura somente nosso arquivo de laboratório.
```

### 🧪 LABORATÓRIO — salt e armazenamento de senha

**Salt** é um valor aleatório individual e público, armazenado junto ao verificador. Ele faz com
que duas contas com a mesma senha tenham verificadores diferentes. Salt não fortalece a senha
escolhida e não precisa ficar secreto. Reutilize o salt salvo para verificar aquela conta.

SHA-256 simples, mesmo com salt, é rápido demais para ser nosso mecanismo de armazenamento
de senhas. Aqui usamos **PBKDF2-HMAC-SHA256**, que repete uma derivação com custo configurável.
O laboratório usa 600.000 iterações; é um parâmetro didático explícito, a calibrar e revisar no
ambiente real. Argon2id é uma alternativa com custo de memória; bcrypt aparece em sistemas
existentes. Não faremos busca de senhas nem quebra de hashes.


```python
ITERACOES_LAB = 600_000

def cadastrar_senha(senha):
    salt = secrets.token_bytes(16)  # Um novo salt para cada cadastro.
    derivado = hashlib.pbkdf2_hmac("sha256", senha.encode("utf-8"), salt, ITERACOES_LAB)
    return {"algoritmo": "pbkdf2-sha256", "iteracoes": ITERACOES_LAB,
            "salt": salt.hex(), "verificador": derivado.hex()}

def verificar_senha(senha, registro):
    # Este registro é interno e confiável, não um objeto arbitrário vindo da rede.
    if registro["algoritmo"] != "pbkdf2-sha256":
        raise ValueError("Algoritmo não suportado")
    derivado = hashlib.pbkdf2_hmac(
        "sha256", senha.encode("utf-8"), bytes.fromhex(registro["salt"]), registro["iteracoes"])
    return hmac.compare_digest(derivado.hex(), registro["verificador"])

senha_ficticia = "Filtro-laranja!Apenas-exemplo"
conta_a = cadastrar_senha(senha_ficticia)
conta_b = cadastrar_senha(senha_ficticia)
print("Salt da conta A:", conta_a["salt"])
print("Salt da conta B:", conta_b["salt"])
confirmar(conta_a["verificador"] != conta_b["verificador"], "Mesma senha, verificadores diferentes")
confirmar(verificar_senha(senha_ficticia, conta_a), "Senha correta aceita")
confirmar(not verificar_senha("Outra senha fictícia", conta_a), "Senha incorreta recusada")
confirmar("senha" not in conta_a, "Registro contém verificador, não a senha em claro")
```

## Bloco 2 — Criptografia Simétrica: AES-GCM

### 📖 CONCEITO

Quem cifra e quem decifra compartilham uma chave secreta. Com AES-GCM, usamos **chave de 256
bits**, **nonce de 12 bytes** e uma **tag de autenticação** incluída no resultado pela biblioteca.
Os **dados associados (AAD)** permanecem legíveis, mas sua alteração também deve ser detectada.

**Nunca repita um nonce com a mesma chave.** Sorteamos nonces neste laboratório de pequeno
volume; isso não equivale a garantir unicidade em sistemas distribuídos de grande escala.

## Bloco 3 — Criptografia Assimétrica: RSA + chaves pública/privada

### 📖 CONCEITO

A oficina gera um par: a **chave pública** pode ser distribuída; a **privada** fica sob sua guarda.
A equipe de campo cifra para a chave pública da oficina, que decifra com a privada.
Usaremos RSA de 2048 bits e **OAEP com SHA-256**, não RSA sem preenchimento.

**PEM** é um formato textual para transportar chaves. A privada será serializada em PKCS8
com proteção por senha; a pública, em SubjectPublicKeyInfo. A senha deste exercício é
aleatória e fica apenas em memória, sem impressão nem gravação em arquivo.

### 🧪 LABORATÓRIO — geração e ida e volta em PEM


```python
def gerar_par_rsa():
    privada = rsa.generate_private_key(public_exponent=65537, key_size=2048)
    return privada, privada.public_key()

def oaep_sha256():
    return padding.OAEP(mgf=padding.MGF1(hashes.SHA256()), algorithm=hashes.SHA256(), label=None)

privada_oficina, publica_oficina = gerar_par_rsa()
senha_pem = secrets.token_urlsafe(32).encode("ascii")
pem_privado = privada_oficina.private_bytes(
    encoding=serialization.Encoding.PEM,
    format=serialization.PrivateFormat.PKCS8,
    encryption_algorithm=serialization.BestAvailableEncryption(senha_pem))
pem_publico = publica_oficina.public_bytes(
    encoding=serialization.Encoding.PEM,
    format=serialization.PublicFormat.SubjectPublicKeyInfo)

# Não mostramos o conteúdo privado. O objeto recarregado é usado no próximo laboratório.
privada_recarregada = serialization.load_pem_private_key(pem_privado, password=senha_pem)
publica_recarregada = serialization.load_pem_public_key(pem_publico)
print("Formato privado:", pem_privado.splitlines()[0].decode("ascii"))
print("Formato público:", pem_publico.splitlines()[0].decode("ascii"))
confirmar(publica_recarregada.public_numbers() == publica_oficina.public_numbers(),
          "Chave pública preservada após serialização")
esperar_erro(ValueError,
             lambda: serialization.load_pem_private_key(pem_privado, password=b"senha-errada"),
             "PEM privado protegido recusa senha incorreta")
```

```python
segredo_curto = "Código interno fictício: manutenção verde".encode("utf-8")
cifra_rsa = publica_recarregada.encrypt(segredo_curto, oaep_sha256())
decifrado_rsa = privada_recarregada.decrypt(cifra_rsa, oaep_sha256())
confirmar(decifrado_rsa == segredo_curto, "RSA-OAEP recupera o segredo curto")
print("Mensagem:", len(segredo_curto), "bytes | Cifra RSA:", len(cifra_rsa), "bytes")

# OAEP: máximo = k - 2*hLen - 2, com k e hLen em bytes.
k_bytes = (publica_oficina.key_size + 7) // 8
limite_oaep = k_bytes - 2 * hashes.SHA256().digest_size - 2
print("Limite de entrada neste par RSA/OAEP:", limite_oaep, "bytes")
esperar_erro(ValueError,
             lambda: publica_oficina.encrypt(b"X" * (limite_oaep + 1), oaep_sha256()),
             "RSA-OAEP rejeita mensagem acima do limite")
```

### 🧪 LABORATÓRIO — criptografia híbrida

RSA-OAEP de 2048 bits com SHA-256 aceita até **190 bytes**. Para documentos maiores, usamos
AES-GCM nos dados e RSA-OAEP apenas na chave AES. Este é um exercício de composição, não
um protocolo de transporte completo. A chave pública do destinatário precisa ser confiável.

## Bloco 4 — Assinatura Digital: assinar e validar mensagem

### 📖 CONCEITO

A equipe de campo cria **outro par RSA, exclusivo para assinatura**. Ela assina com a privada;
a oficina verifica com a pública confiável da equipe. Usaremos **RSA-PSS e SHA-256**.

A assinatura vincula os bytes à posse da chave privada; não torna verdadeiro o conteúdo nem
estabelece a identidade do titular sem confiança na chave pública. Não descreva assinatura
como “criptografar com a chave privada”: os esquemas e objetivos são diferentes.

### 🧪 LABORATÓRIO — assinatura, validação e adulteração


```python
def pss_sha256():
    return padding.PSS(mgf=padding.MGF1(hashes.SHA256()), salt_length=padding.PSS.MAX_LENGTH)

def assinar_bytes(privada, dados):
    return privada.sign(dados, pss_sha256(), hashes.SHA256())

def validar_assinatura(publica, dados, assinatura):
    try:
        publica.verify(assinatura, dados, pss_sha256(), hashes.SHA256())
    except InvalidSignature:
        return False
    return True

privada_campo, publica_campo = gerar_par_rsa()
assinatura_boletim = assinar_bytes(privada_campo, boletim_bytes)
confirmar(validar_assinatura(publica_campo, boletim_bytes, assinatura_boletim), "Assinatura original válida")
confirmar(not validar_assinatura(publica_campo, alterar_um_bit(boletim_bytes), assinatura_boletim),
          "Mensagem adulterada invalida assinatura")
confirmar(not validar_assinatura(publica_campo, boletim_bytes, alterar_um_bit(assinatura_boletim)),
          "Assinatura adulterada recusada")
confirmar(not validar_assinatura(publica_oficina, boletim_bytes, assinatura_boletim),
          "Chave pública de outro par não valida a assinatura")
print("Tamanho da assinatura:", len(assinatura_boletim), "bytes")
```
